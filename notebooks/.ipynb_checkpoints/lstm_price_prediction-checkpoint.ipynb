```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/Price_Agriculture_commodities_Week.csv')

print(df.head())
```

```python
commodity_df = df[df['Commodity'] == 'Brinjal']

print(commodity_df.head())
```

```python
commodity_df = df[df['Commodity'] == 'Brinjal']

commodity_df['Arrival_Date'] = pd.to_datetime(
    commodity_df['Arrival_Date'],
    dayfirst=True
)

commodity_df = commodity_df.sort_values('Arrival_Date')

price_series = commodity_df.groupby(
    'Arrival_Date'
)['Modal Price'].mean()

print(price_series.head())

print(len(price_series))
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(
    np.array(price_series).reshape(-1,1)
)

print(scaled_data[:5])
```

```python
sequence_length = 3

X = []
y = []

for i in range(sequence_length, len(scaled_data)):

    X.append(
        scaled_data[i-sequence_length:i]
    )

    y.append(
        scaled_data[i]
    )

X = np.array(X)
y = np.array(y)

print(X.shape)
print(y.shape)
```

```python
print(df['Commodity'].unique())
```

```python
import torch
import torch.nn as nn

X_tensor = torch.tensor(X, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.float32)

print(X_tensor.shape)
print(y_tensor.shape)
```

```python
class LSTMModel(nn.Module):

    def __init__(self):
        super(LSTMModel, self).__init__()

        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=50,
            batch_first=True
        )

        self.fc = nn.Linear(50, 1)

    def forward(self, x):

        out, _ = self.lstm(x)

        out = out[:, -1, :]

        out = self.fc(out)

        return out
```

```python
model = LSTMModel()

criterion = nn.MSELoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
epochs = 100

for epoch in range(epochs):

    outputs = model(X_tensor)

    loss = criterion(outputs, y_tensor)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if (epoch + 1) % 10 == 0:
        print(
            f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}'
        )
```

```python
last_sequence = scaled_data[-sequence_length:]

last_sequence = torch.tensor(
    last_sequence,
    dtype=torch.float32
).unsqueeze(0)

prediction = model(last_sequence)

predicted_price = scaler.inverse_transform(
    prediction.detach().numpy()
)

print(
    "Predicted Brinjal Price:",
    predicted_price[0][0]
)
```

```python
import matplotlib.pyplot as plt
```

```python
predictions = model(X_tensor).detach().numpy()

predictions = scaler.inverse_transform(predictions)

actual = scaler.inverse_transform(
    y.reshape(-1,1)
)

plt.figure(figsize=(10,5))

plt.plot(actual, label='Actual Price')

plt.plot(predictions, label='Predicted Price')

plt.legend()

plt.title(
    "Actual vs Predicted Prices"
)

plt.show()
```

```python
import torch
import torch.nn as nn
import numpy as np
```

```python
class LSTMModel(nn.Module):

    def __init__(self):
        super(LSTMModel, self).__init__()

        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=50,
            num_layers=1,
            batch_first=True
        )

        self.fc = nn.Linear(50, 1)

    def forward(self, x):

        out, _ = self.lstm(x)

        out = self.fc(out[:, -1, :])

        return out
```

```python
model = LSTMModel()

model.load_state_dict(
    torch.load('brinjal_lstm_model.pth')
)

model.eval()
```

```python
future_input = torch.tensor(
    X[-1:].astype(np.float32)
)

prediction = model(future_input)

prediction = prediction.detach().numpy()

predicted_price = scaler.inverse_transform(
    prediction
)

print(predicted_price)
```

```python
commodities = [
    'Brinjal',
    'Cabbage',
    'Cauliflower',
    'Rice',
    'Wheat'
]
```

```python
for commodity in commodities:

    print(f"Training for {commodity}")

    commodity_df = df[df['Commodity'] == commodity]

    commodity_df['Arrival_Date'] = pd.to_datetime(
        commodity_df['Arrival_Date'],
        dayfirst=True
    )

    commodity_df = commodity_df.sort_values(
        'Arrival_Date'
    )

    price_series = commodity_df.groupby(
        'Arrival_Date'
    )['Modal Price'].mean()

    if len(price_series) < 5:

        print(f"Not enough data for {commodity}")
        continue

    scaler = MinMaxScaler()

    scaled_data = scaler.fit_transform(
        np.array(price_series).reshape(-1,1)
    )

    sequence_length = 3

    X = []
    y = []

    for i in range(sequence_length, len(scaled_data)):

        X.append(
            scaled_data[i-sequence_length:i]
        )

        y.append(
            scaled_data[i]
        )

    X = np.array(X)
    y = np.array(y)

    X_tensor = torch.tensor(
        X.astype(np.float32)
    )

    y_tensor = torch.tensor(
        y.astype(np.float32)
    )

    model = LSTMModel()

    criterion = nn.MSELoss()

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=0.01
    )

    epochs = 100

    for epoch in range(epochs):

        outputs = model(X_tensor)

        loss = criterion(outputs, y_tensor)

        optimizer.zero_grad()

        loss.backward()

        optimizer.step()

    filename = f"{commodity.lower()}_lstm_model.pth"

    torch.save(
        model.state_dict(),
        filename
    )

    print(f"{commodity} model saved")
```
